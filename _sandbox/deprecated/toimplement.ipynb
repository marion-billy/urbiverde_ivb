```python
import geopandas as gpd
import ee
import geemap
from utils_raster import *
from utils_vector import *
class AOISetup:
    """
    Classe pour préparer et stocker les différentes représentations d'une AOI.
    """

    def __init__(self, aoi_raw: gpd.GeoDataFrame):
        """
        Initialise la classe avec le GeoDataFrame brut et lance la préparation.
        
        Args:
            aoi_raw (gpd.GeoDataFrame): Polygone de la zone d'étude en coordonnées géographiques.
        """
        self.aoi_raw = aoi_raw
        self.utm_epsg = None
        self.aoi_utm = None
        self.aoi_ee = None

        self._prepare()

    def _prepare(self):
        """Prépare toutes les versions de l'AOI."""
        self.utm_epsg = str(aoi_raw.estimate_utm_crs())
        self.aoi_utm = self.aoi_raw.to_crs(self.utm_epsg)
        self.aoi_ee = geemap.gdf_to_ee(gdf_to_bbox(self.aoi_utm))

    def get_outputs(self) -> tuple[gpd.GeoDataFrame, ee.Geometry, str]:
        """
        Retourne les objets préparés.
        
        Returns:
            tuple: (aoi_utm, aoi_ee, utm_epsg)
        """
        return self.aoi_utm, self.aoi_ee, self.utm_epsg
```
